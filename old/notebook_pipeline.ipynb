## 3.3 -- Drift Detection
We compare our reference split (Train: 2024-01 & 2024-03) against the current split (Test: 2025-01) to identify data drift (PSI + KS test), label drift, and concept drift.

```python
print("--- Running Drift Detection ---")

# 1. Feature Data Drift
drift_report = build_drift_report(
    reference_df=df_clean_train, 
    current_df=df_clean_test, 
    feature_cols=["trip_distance", "PULocationID", "DOLocationID"]
)
display(drift_report)

# 2. Target Label Drift
label_drift = detect_label_drift(df_clean_train, df_clean_test)
print("\nLabel Drift metrics:")
print(label_drift)

# 3. Concept Drift
concept_drift = detect_concept_drift(
    reference_raw_df=df_clean_train, 
    current_raw_df=df_clean_test, 
    model=champion_model, 
    scaler=eng_scaler
)
print("\nConcept Drift metrics:")
print(concept_drift)

# Plot distributions
plot_feature_distributions(df_clean_train, df_clean_test, "trip_distance", cur_label="Jan 2025 Test", output_dir=PLOTS_DIR)
plot_label_drift_distribution(df_clean_train, df_clean_test, cur_label="Jan 2025 Test", output_dir=PLOTS_DIR)
```

## 3.4 -- Drift Mitigation
We apply drift mitigation (recalibrating the scaler on recent data) and compare the performance of the original vs. mitigated model.

```python
print("--- Running Drift Mitigation ---")

# Run Recalibrate mitigation strategy
mitigated_model, mitigated_scaler, eval_steps = mitigate(
    strategy="recalibrate",
    train_df=df_clean_train,
    recent_df=df_clean_test,
    model_name="gradient_boosting",
    model_dir="models/mitigated",
    base_model=champion_model
)

# Compute prediction errors for comparison
y_pred_original = champion_model.predict(X_test_eng)
original_errors = np.abs(y_test_eng - y_pred_original)

# Apply new scaler to test features
X_test_recalibrated = X_test_eng.copy()
scale_cols = [c for c in mitigated_scaler.feature_names_in_ if c in X_test_recalibrated.columns]
X_test_recalibrated[scale_cols] = mitigated_scaler.transform(X_test_recalibrated[scale_cols])

y_pred_mitigated = champion_model.predict(X_test_recalibrated)
mitigated_errors = np.abs(y_test_eng - y_pred_mitigated)

error_arrays = {
    "Original Model (Jan 2025 Test)": original_errors,
    "Mitigated Model (Recalibrate)": mitigated_errors
}

plot_mitigation_comparison(error_arrays, output_dir=PLOTS_DIR)
```

## 4.1 -- Evidently AI Drift Detection (Train vs. Dec 2024)
We compare our baseline training features against the December 2024 dataset using Evidently AI. This detects shifts in feature distributions (dataset drift) and performance degradation (concept drift), choosing the best mitigation strategy.

```python
import pandas as pd
from pathlib import Path
from src.drift_detection_evidently import (
    run_evidently_drift_report, parse_drift_results,
    run_evidently_concept_drift_report, parse_concept_drift_results,
    select_mitigation_strategy
)
from src.features import run_feature_pipeline, TARGET_COL
from src.cleaning import clean_dataframe

DRIFT_RAW_PARQUET = "data/yellow_tripdata_2024-12.parquet"
drift_month_path = Path(DRIFT_RAW_PARQUET)

if not drift_month_path.exists():
    print(f"Drift month parquet not found at {DRIFT_RAW_PARQUET}. Place the file there to run Evidently analysis.")
else:
    print("Loading and cleaning December 2024 drift month...")
    drift_df = pd.read_parquet(DRIFT_RAW_PARQUET)
    drift_df = clean_dataframe(drift_df, is_train=True)
    drift_df["tpep_pickup_datetime"] = pd.to_datetime(drift_df["tpep_pickup_datetime"])

    # Split: Dec 1-21 for training mitigation, Dec 22-31 for evaluation
    DRIFT_SEED = 42
    drift_train_raw = drift_df[drift_df["tpep_pickup_datetime"].dt.day <= 21].sample(20000, random_state=DRIFT_SEED).reset_index(drop=True)
    drift_eval_raw  = drift_df[drift_df["tpep_pickup_datetime"].dt.day >= 22].sample(5000, random_state=DRIFT_SEED).reset_index(drop=True)
    print(f"  Mitigation train set : {len(drift_train_raw):,} rows")
    print(f"  Mitigation eval set  : {len(drift_eval_raw):,} rows")

    # Engineer features with original scaler
    drift_train_eng, _ = run_feature_pipeline(drift_train_raw, scaler=eng_scaler, is_training=False)
    drift_eval_eng,  _ = run_feature_pipeline(drift_eval_raw,  scaler=eng_scaler, is_training=False)

    # Reference df (Jan training engineered)
    ref_eng_df = X_train_eng.copy()
    ref_eng_df[TARGET_COL] = y_train_eng.values

    # 1. Run Evidently Dataset & Target Drift
    print("\nRunning Evidently dataset + label drift report...")
    evidently_report = run_evidently_drift_report(ref_eng_df, drift_train_eng)
    drift_results = parse_drift_results(evidently_report)

    print(f"  Overall drift detected : {drift_results['overall_drift']}")
    print(f"  Features drifted       : {drift_results['n_drifted']} ({drift_results['share_drifted']:.1%} of columns)")
    if drift_results["drifted_features"]:
        print(f"  Drifted feature names  : {drift_results['drifted_features']}")

    # Save HTML report
    Path("outputs").mkdir(exist_ok=True)
    evidently_html = Path("outputs") / "evidently_drift_report.html"
    evidently_report.save_html(str(evidently_html))
    print(f"  Evidently report HTML saved -> {evidently_html}")

    # 2. Run Evidently Concept Drift
    print("\nRunning Evidently concept drift report...")
    ref_perf_df = X_test_eng.copy()
    ref_perf_df[TARGET_COL] = y_test_eng.values
    ref_perf_df["prediction"] = champion_model.predict(X_test_eng)

    cur_perf_df = drift_eval_eng.copy()
    cur_perf_df["prediction"] = champion_model.predict(drift_eval_eng.drop(columns=[TARGET_COL]))

    concept_drift_report = run_evidently_concept_drift_report(ref_perf_df, cur_perf_df)
    concept_drift_results = parse_concept_drift_results(concept_drift_report)

    print(f"  Concept drift detected : {concept_drift_results['concept_drift_detected']}")
    print(f"  Reference MAE (Jan)    : ${concept_drift_results['ref_mae']:.2f}")
    print(f"  Current MAE (Dec eval) : ${concept_drift_results['cur_mae']:.2f}")

    concept_drift_html = Path("outputs") / "evidently_concept_drift_report.html"
    concept_drift_report.save_html(str(concept_drift_html))
    print(f"  Evidently concept drift HTML saved -> {concept_drift_html}")

    # 3. Select optimal mitigation strategy
    selected_strategy = select_mitigation_strategy(drift_results, concept_drift_results)
    print(f"\n  Selected strategy: {selected_strategy}")
```

## 4.2 -- Drift Mitigation & W&B Artifact Versioning
We apply the mitigation strategy chosen by Evidently, evaluate the performance recovery on our evaluation split, plot the error comparison, and version our data, feature engineering (scaler), and mitigated model in W&B.

```python
if 'selected_strategy' in locals() and selected_strategy != "none":
    from src.drift_mitigation import mitigate, plot_mitigation_comparison
    from src.experiment_tracking import ExperimentTracker
    from src.versioning import log_data_artifact, log_model_artifact, log_feature_artifact
    import numpy as np
    import joblib

    print(f"Applying mitigation strategy: {selected_strategy}...")
    mitigated_model, mitigated_scaler, eval_steps = mitigate(
        strategy=selected_strategy,
        train_df=df_clean_train,
        recent_df=drift_train_raw,
        model_name="gradient_boosting",
        model_dir="models/mitigated",
        base_model=champion_model,
        drifted_features=drift_results["drifted_features"]
    )

    if mitigated_model is None:
        mitigated_model = champion_model

    # Re-engineer evaluation set using mitigated scaler
    active_scaler = mitigated_scaler if mitigated_scaler is not None else eng_scaler
    drift_eval_eng_mit, _ = run_feature_pipeline(
        drift_eval_raw, scaler=active_scaler, is_training=False, custom_creation_steps=eval_steps
    )

    # Evaluate performance
    y_drift_eval = drift_eval_eng[TARGET_COL].values
    y_pred_drift_base = champion_model.predict(drift_eval_eng.drop(columns=[TARGET_COL]))
    y_pred_drift_mit  = mitigated_model.predict(drift_eval_eng_mit.drop(columns=[TARGET_COL]))

    baseline_drift_mae = float(np.mean(np.abs(y_drift_eval - y_pred_drift_base)))
    mitigated_drift_mae = float(np.mean(np.abs(y_drift_eval - y_pred_drift_mit)))
    improvement = (baseline_drift_mae - mitigated_drift_mae) / baseline_drift_mae * 100

    print(f"\nEvaluation comparison:")
    print(f"  Pre-mitigation MAE  : ${baseline_drift_mae:.2f}")
    print(f"  Post-mitigation MAE : ${mitigated_drift_mae:.2f}")
    print(f"  Improvement         : {improvement:+.1f}%")

    # Plot comparison
    comparison_fig = plot_mitigation_comparison(
        {
            "Champion - Jan (In-Dist)": np.abs(y_test_eng.values - champion_model.predict(X_test_eng)),
            "Champion - Dec (Drifted)": np.abs(y_drift_eval - y_pred_drift_base),
            f"Mitigated ({selected_strategy}) - Dec": np.abs(y_drift_eval - y_pred_drift_mit)
        },
        output_dir=PLOTS_DIR
    )

    # Log to W&B
    mit_tracker = ExperimentTracker(
        project=WANDB_PROJECT,
        run_name=f"mitigation-{selected_strategy}-dec",
        tags=["drift-mitigation", "december", selected_strategy],
        config={
            "strategy": selected_strategy,
            "drifted_features": drift_results["drifted_features"]
        }
    )
    mit_tracker.log_summary({
        "baseline_drift_mae": baseline_drift_mae,
        "mitigated_drift_mae": mitigated_drift_mae,
        "mae_improvement_pct": improvement
    })
    mit_tracker.log_plot(comparison_fig, "mitigation_comparison")

    # Version the evaluation set
    drift_eval_parquet = Path("data/processed/drift_eval.parquet")
    drift_eval_raw.to_parquet(drift_eval_parquet, index=False)
    log_data_artifact(mit_tracker, drift_eval_parquet, "december-eval-set", metadata={"month": "December 2024"})

    # Version mitigated model & scaler
    mitigated_scaler_path = Path("models/mitigated/scaler_mitigated.pkl")
    joblib.dump(active_scaler, mitigated_scaler_path)

    log_feature_artifact(
        mit_tracker, mitigated_scaler_path,
        active_feature_steps=[c for c in drift_eval_eng_mit.columns if c != TARGET_COL]
    )

    if selected_strategy in ("reweight_retrain", "drop_features"):
        mit_pkl_name = "gradient_boosting_reweighted.pkl" if selected_strategy == "reweight_retrain" else "gradient_boosting_drop_features.pkl"
        log_model_artifact(mit_tracker, Path("models/mitigated") / mit_pkl_name, "mitigated-model")

    mit_tracker.log_artifact(evidently_html, "evidently-drift-report", "report")
    run_url = mit_tracker.finish()
    if run_url:
        print(f"\nMitigation run logged to W&B -> {run_url}")
else:
    print("No mitigation applied (strategy is 'none' or missing).")
```
